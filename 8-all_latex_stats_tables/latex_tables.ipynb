```python
"""
Three Summary Tables
=====================
Table 1: Open source product adoption (from MAIN.csv)
Table 2: Library risk by name (from general_library_risk_by_name.csv)
Table 3: Vulnerable library impact (from vulnerable_library_impact.csv)

Each table: top 10 rows sorted as specified + average row at the bottom.
"""

import pandas as pd
import numpy as np

# ── File paths ────────────────────────────────────────────────────────────────
MAIN_PATH    = "/Users/dmk6603/Documents/swdb_opensource/5.2-merge_counts/MAIN.csv"
BYNAME_PATH  = "/Users/dmk6603/Documents/swdb_opensource/general_library_risk_by_name.csv"
IMPACT_PATH  = "/Users/dmk6603/Documents/swdb_opensource/vulnerable_library_impact.csv"

# ── Helpers ───────────────────────────────────────────────────────────────────
def clean_numeric(series):
    """Strip commas and cast to numeric."""
    return (
        series.astype(str)
        .str.replace(",", "", regex=False)
        .str.strip()
        .pipe(pd.to_numeric, errors="coerce")
    )

def average_row(df, label_col, numeric_cols):
    """Build an AVERAGE row for the given numeric columns."""
    avg = {col: df[col].mean() for col in numeric_cols}
    avg[label_col] = "⌀ AVERAGE"
    return pd.DataFrame([avg])

def format_table(df, numeric_cols, pct_cols=None, int_cols=None):
    """
    Format numbers for display:
      - pct_cols  → 2 decimal places + %
      - int_cols  → integer with thousands separator
      - others    → 2 decimal places
    The last row (average) is formatted the same way.
    """
    pct_cols  = pct_cols  or []
    int_cols  = int_cols  or []
    df = df.copy()
    for col in numeric_cols:
        if col in pct_cols:
            df[col] = df[col].apply(
                lambda x: f"{x:.2f}%" if pd.notna(x) else "—"
            )
        elif col in int_cols:
            df[col] = df[col].apply(
                lambda x: f"{int(round(x)):,}" if pd.notna(x) else "—"
            )
        else:
            df[col] = df[col].apply(
                lambda x: f"{x:,.2f}" if pd.notna(x) else "—"
            )
    return df

def print_table(title, df):
    print(f"\n{'═'*100}")
    print(f"  {title}")
    print(f"{'═'*100}")
    print(df.to_string(index=False))
    print()

# ══════════════════════════════════════════════════════════════════════════════
# TABLE 1 — Open source product adoption (MAIN.csv)
# ══════════════════════════════════════════════════════════════════════════════
print("Loading MAIN.csv...")
main = pd.read_csv(MAIN_PATH)
main.columns = main.columns.str.strip()

T1_COLS = [
    "Product", "Total Enterprises", "gh_stars", "gh_forks",
    "total_libraries", "unique_libraries",
    "vuln_lib_count", "pct_libs_vulnerable", "total_vulns"
]

# Parse numeric columns
T1_NUM = [c for c in T1_COLS if c != "Product"]
for col in T1_NUM:
    main[col] = clean_numeric(main[col])

t1 = main[T1_COLS].copy().dropna(subset=["pct_libs_vulnerable"])
t1 = t1.sort_values("pct_libs_vulnerable", ascending=False)

top10_t1 = t1.head(10).copy()
avg_t1   = average_row(top10_t1, "Product", T1_NUM)  # avg of top 10
full_avg_t1 = average_row(t1, "Product", T1_NUM)      # avg of all products
full_avg_t1["Product"] = "⌀ AVG (all products)"

t1_display = pd.concat([top10_t1, full_avg_t1], ignore_index=True)
t1_display = format_table(
    t1_display,
    numeric_cols=T1_NUM,
    pct_cols=["pct_libs_vulnerable"],
    int_cols=["Total Enterprises", "gh_stars", "gh_forks",
              "total_libraries", "unique_libraries",
              "vuln_lib_count", "total_vulns"]
)

print_table("TABLE 1 — Open Source Product Adoption (sorted by % Vulnerable Libraries)", t1_display)

# ══════════════════════════════════════════════════════════════════════════════
# TABLE 2 — Library risk by name
# ══════════════════════════════════════════════════════════════════════════════
print("Loading general_library_risk_by_name.csv...")
byname = pd.read_csv(BYNAME_PATH)
byname.columns = byname.columns.str.strip()

T2_COLS = ["Library Name", "Total Products Using It", "Total Enterprises Exposed", "Total CVEs"]
T2_NUM  = ["Total Products Using It", "Total Enterprises Exposed", "Total CVEs"]

for col in T2_NUM:
    byname[col] = clean_numeric(byname[col])

t2 = byname[T2_COLS].copy().dropna(subset=["Total Products Using It"])
t2 = t2.sort_values("Total Products Using It", ascending=False)

top10_t2    = t2.head(15).copy()
full_avg_t2 = average_row(t2, "Library Name", T2_NUM)
full_avg_t2["Library Name"] = "⌀ AVG (all libraries)"

t2_display = pd.concat([top10_t2, full_avg_t2], ignore_index=True)
t2_display = format_table(
    t2_display,
    numeric_cols=T2_NUM,
    int_cols=["Total Products Using It", "Total Enterprises Exposed", "Total CVEs"]
)

print_table("TABLE 2 — Library Risk by Name (sorted by Total Products Using It)", t2_display)

# ══════════════════════════════════════════════════════════════════════════════
# TABLE 3 — Vulnerable library impact
# ══════════════════════════════════════════════════════════════════════════════
print("Loading vulnerable_library_impact.csv...")
impact = pd.read_csv(IMPACT_PATH)
impact.columns = impact.columns.str.strip()

# Rename column if it has a leading space
impact.columns = [c.strip() for c in impact.columns]

T3_COLS = [
    "Vulnerable Library", "CVE ID", "CVE Severity",
    "Products Affected", "Enterprises Exposed"
]
T3_NUM  = ["Products Affected", "Enterprises Exposed"]

for col in T3_NUM:
    impact[col] = clean_numeric(impact[col])

t3 = impact[T3_COLS].copy().dropna(subset=["Products Affected"])
t3 = t3.sort_values("Products Affected", ascending=False)

top10_t3    = t3.head(15).copy()
full_avg_t3 = average_row(t3, "Vulnerable Library", T3_NUM)
full_avg_t3["Vulnerable Library"] = "⌀ AVG (all CVEs)"
full_avg_t3["CVE ID"]             = ""
full_avg_t3["CVE Severity"]       = ""

t3_display = pd.concat([top10_t3, full_avg_t3], ignore_index=True)
t3_display = format_table(
    t3_display,
    numeric_cols=T3_NUM,
    int_cols=["Products Affected", "Enterprises Exposed"]
)

print_table("TABLE 3 — Vulnerable Library Impact (sorted by Products Affected)", t3_display)

print("✅ Done.")
```
